# TABULATE

```python
from tabulate import tabulate
name=input("enter your name")
age=input("enter age")
project_name=input("enter project name")
address=input("enter address")

data=[
     ["name",name],
     ["age",age],
     ["project",project_name],
     ["address",address]
]
print(tabulate(data,headers=["field","details"],tablefmt="grid"))
```

# loop statements (while)
while

```python
num=1;
while num<=100:
    if num%2==0:
        print(num)
    num+=1;
```

# for loop

```python
#prime no
for i in range(2,100):
    count=0
    for j in range(1,100):
        if i%j==0:
            count+=1
    if count==2:
        print(i)
```

```python
#to print the table ( a x b = c )
a= int(input("enter value to find its multiple"))
# for loop executes in the pattern "for i in range(init, n-1) 
for i in range(1,10):
    print(f" {a} x {i} = {a*i}")
```
