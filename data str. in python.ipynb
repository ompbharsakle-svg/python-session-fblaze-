## predefine methods of list 

```python
#.append()
data=[1,2,"a","b","cd"]
#     1 2  3   4   5  
print("initial =",data)
data.append(30)
print(data)

#insert
data.insert(3,"abc")
print(data)

#pop() ==> remove last element if index is not given
data.pop() 
print(data)
data.pop(3) 
print(data)
```

# set

## or operator

```python
my_set={29,46,47}
my_set= my_set | {40}
print(my_set)
```

# negation

```python
my_set={29,46,47}
my_set=my_set -{47}
print(my_set)
```

```python
my_set1={29,46,47}
my_set2={30,40,4}
myset=my_set1 | my_set2
print(myset)
```

# and oper.

```python
my_set1={29,46,47}
my_set2={30,46,29,4}
myset=my_set1 & my_set2
print(myset)
```

# .union()

```python
my_set1={29,46,47}
my_set2={47,4,29,4}
print(my_set2.union(my_set1))
```

# .add()

```python
my_set={29,46,47}
my_set.add(40)
print(myset)
```

# .intersection()

```python
my_set1={29,46,47}
my_set2={30,46,29,4}
print(my_set1.intersection(my_set2))
```

# symetric_difference

```python
my_set1={29,46,47}
my_set2={30,46,29,4}
print(my_set1.symmetric_difference(my_set2))
```
