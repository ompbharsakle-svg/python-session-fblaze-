# list

```python
my_data=[1,2,3,4,5]
print(my_data)
print(type(my_data))
print(my_data[1])
#slicing
print(my_data[0:2])
#updating
my_data.insert(3,100)
#my_data[3]=100;
print(my_data)
#adding element to end of the list
my_data +=[6]
print(my_data)
```

## predefine methods of list 

```python
#.append()
data=[1,2,"a","b","cd"]
#     1 2  3   4   5  
print("initial =",data)
data.append(30)
print(data)

#insert
data.insert(3,"abc")
print(data)

#pop() ==> remove last element if index is not given
data.pop() 
print(data)
data.pop(3) 
print(data)
```

# sort()

```python
data=[2,4,1,6,4,7]
data.sort()
print(data)
```

# reverse() 

```python
data=[2,4,1,6,4,7]
data.reverse()
print(data)
```

# len()

```python
data=[2,1,4,4,1,6,4,7]
print(len(data))
```

# count()

```python
data=[2,1,4,4,1,6,4,7]
print(data.count(1))
```

# extend()

```python
data=[2,1,4,4,1,6,4,7]
data.extend([466,583,578])
print(data)
```

```python
data=[2,1,4,4,1,6,4,7]
data2=[466,583,578]
#my data is already defined in previous codes 
my_data.extend(data)
print(my_data)
my_data.extend(data2)
print(my_data)
```

# max(),min(),sum()

```python
print("maximum=",max(my_data))
print("minimum=",min(my_data))
print("sum=",sum(my_data))
```

# slicing in tuple

```python
my_tuple=(1,4,2,5,7,2)
print(my_tuple)
a=my_tuple[1:4]
print(a)
```

# conversion of tuple into list

```python
my_tuple=(1,4,2,5,7,2)
print(type(my_tuple))
b=list(my_tuple)
print(type(b))
b.append(67)
print(b)
```

# count(), index() in tuple

```python
#.count()
my_tuple=(1,4,2,5,7,2)
print(my_tuple.count(5))

#.index()
print(my_tuple.index(5))
```
